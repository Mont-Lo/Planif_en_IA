# Importation et définition de variables globales

```python
import gymnasium as gym
import numpy as np
import torch
import random

from model import nn_rainbow
```

```python
# environment
env = gym.make("ALE/Freeway-v5", render_mode="rgb_array")
```

## Set random seed

```python
seed = 777

def seed_torch(seed):
    torch.manual_seed(seed)
    if torch.backends.cudnn.enabled:
        torch.cuda.manual_seed(seed)
        torch.backends.cudnn.benchmark = False
        torch.backends.cudnn.deterministic = True

np.random.seed(seed)
random.seed(seed)
seed_torch(seed)
```

# Exemple d'execution de rainbow (même format pour ses variantes)

```python
# parameters
num_frames = 10000
memory_size = 10000
batch_size = 128
target_update = 100
```

```python
agent = nn_rainbow.DQNAgent(env, memory_size, batch_size, target_update, seed)
```

```python
# train
agent.train(num_frames)
```

```python
# test
video_folder="videos_rainbow/"
agent.test(video_folder=video_folder)
```
